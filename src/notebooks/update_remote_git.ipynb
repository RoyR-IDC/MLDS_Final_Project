# Update remote git

```python
import subprocess
import os
from google.colab import drive # only available in Colab, so this import will fail if not in Colab, which is what we want to detect Colab environment
```

```python
# mount drive
drive.mount('/content/drive')

# get latest code from git
path_to_repo = '/content/drive/MyDrive/MLDS_Final_Project'
```

```python
def update_git_repo():
    try:
        os.chdir(path_to_repo)
        result = subprocess.run(['git', 'pull'], check=True, capture_output=True, text=True)
        print("Update successful:", result.stdout)
    except subprocess.CalledProcessError as e:
        print("Git pull failed!")
        print("Error details:", e.stderr)
        raise Exception("Stopping execution due to Git Pull failure")

update_git_repo() # happens elsewhsr
```
